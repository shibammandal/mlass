# Assignment 3

Perform linear regression with one feature using gradient descent (without using library function) to predict the salary of an employee based on the feature Years Experience. Use hyper-parameter tuning for the best result. Plot the hypothesis function and the data points after each epoch. Evaluate the accuracy of the models by computing relevant performance metrics, including the R^2 value.
Dataset: salary_dataset.csv

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Load Real Dataset
df = pd.read_csv('salary_dataset.csv')
x = df['YearsExperience'].values.astype(float)
y = df['Salary'].values.astype(float)
n = float(len(x))

# 2. Gradient Descent Algorithm with Learning Rate Tuning
def train_gd(x, y, lr=0.01, epochs=2000):
    w, b = 0.0, 0.0
    history = []
    for epoch in range(1, epochs + 1):
        y_pred = w * x + b
        dw = (-2.0 / n) * np.sum(x * (y - y_pred))
        db = (-2.0 / n) * np.sum(y - y_pred)
        w -= lr * dw
        b -= lr * db
        if epoch % (epochs // 10) == 0 or epoch == 1:
            cost = (1.0 / n) * np.sum((y - (w * x + b)) ** 2)
            history.append((epoch, w, b, cost))
    return w, b, history

# Hyperparameter search over learning rates
lrs = [0.001, 0.01, 0.02]
best_w, best_b, best_r2, best_lr, best_hist = 0, 0, -np.inf, None, None
for lr in lrs:
    w, b, hist = train_gd(x, y, lr=lr, epochs=5000)
    pred = w * x + b
    r2 = 1.0 - (np.sum((y - pred) ** 2) / np.sum((y - np.mean(y)) ** 2))
    if r2 > best_r2:
        best_w, best_b, best_r2, best_lr, best_hist = w, b, r2, lr, hist

# 3. Compute Metrics
y_final_pred = best_w * x + best_b
mse = np.mean((y - y_final_pred) ** 2)
mae = np.mean(np.abs(y - y_final_pred))

print("=== Real Data: Gradient Descent from Scratch ===")
print(f"Optimal Learning Rate: {best_lr} | Epochs: 5000")
print(f"Model: Salary = {best_w:.2f} * Experience + {best_b:.2f}")
print(f"MAE: ${mae:.2f} | MSE: {mse:.2f} | R^2: {best_r2:.4f}")

# 4. Plot Hypothesis Function Progression over Epochs
plt.figure(figsize=(8, 4.2))
plt.scatter(x, y, color='darkblue', s=60, edgecolors='k',
            label='Actual Data Points')
x_line = np.linspace(x.min(), x.max(), 100)
for epoch, w_e, b_e, _ in best_hist[::2]:
    plt.plot(x_line, w_e * x_line + b_e, '--', alpha=0.45,
             label=f'Epoch {epoch}')
plt.plot(x_line, best_w * x_line + best_b, 'r-', lw=2.5,
         label=f'Final Fit (R^2={best_r2:.4f})')
plt.xlabel('Years of Experience'); plt.ylabel('Salary ($)')
plt.title('Hypothesis Function Evolution via Gradient Descent')
plt.legend(); plt.grid(True, linestyle=':'); plt.tight_layout()
plt.savefig('gradient_descent_hypothesis.png', dpi=300); plt.show()
```
